```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv(r'C:\BFSI_Loan_Analytics\data\processed\lending_club_features.csv', low_memory=False)

print("Shape:", df.shape)
print("Nulls:", df.isnull().sum().sum())
```

```python
# Chi-Square Test: Loan Grade vs Default
contingency_grade = pd.crosstab(df['grade'], df['loan_outcome'])
print("Contingency Table — Grade vs Default:")
print(contingency_grade)

chi2, p_value, dof, expected = stats.chi2_contingency(contingency_grade)

print(f"\nChi-Square Statistic: {chi2:,.2f}")
print(f"Degrees of Freedom: {dof}")
print(f"P-Value: {p_value}")
print(f"\nConclusion: {'SIGNIFICANT' if p_value < 0.05 else 'NOT SIGNIFICANT'} at 95% confidence level")
```

```python
# Chi-Square Test: Loan Purpose vs Default
contingency_purpose = pd.crosstab(df['purpose'], df['loan_outcome'])
print("Contingency Table — Purpose vs Default:")
print(contingency_purpose)

chi2_p, p_value_p, dof_p, expected_p = stats.chi2_contingency(contingency_purpose)

print(f"\nChi-Square Statistic: {chi2_p:,.2f}")
print(f"Degrees of Freedom: {dof_p}")
print(f"P-Value: {p_value_p}")
print(f"\nConclusion: {'SIGNIFICANT' if p_value_p < 0.05 else 'NOT SIGNIFICANT'} at 95% confidence level")
```

```python
# Split into two groups
income_defaulted = df[df['loan_outcome'] == 1]['annual_inc']
income_paid = df[df['loan_outcome'] == 0]['annual_inc']

print(f"Defaulted group — Mean income: ${income_defaulted.mean():,.2f} | N: {len(income_defaulted):,}")
print(f"Paid group      — Mean income: ${income_paid.mean():,.2f} | N: {len(income_paid):,}")

# Independent T-Test
t_stat, p_value = stats.ttest_ind(income_defaulted, income_paid, equal_var=False)

print(f"\nT-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_value}")
print(f"\nConclusion: {'SIGNIFICANT' if p_value < 0.05 else 'NOT SIGNIFICANT'} at 95% confidence level")
```

```python
# Split into two groups
dti_defaulted = df[df['loan_outcome'] == 1]['dti']
dti_paid = df[df['loan_outcome'] == 0]['dti']

print(f"Defaulted group — Mean DTI: {dti_defaulted.mean():.2f} | N: {len(dti_defaulted):,}")
print(f"Paid group      — Mean DTI: {dti_paid.mean():.2f} | N: {len(dti_paid):,}")

# Independent T-Test
t_stat_dti, p_value_dti = stats.ttest_ind(dti_defaulted, dti_paid, equal_var=False)

print(f"\nT-Statistic: {t_stat_dti:.4f}")
print(f"P-Value: {p_value_dti}")
print(f"\nConclusion: {'SIGNIFICANT' if p_value_dti < 0.05 else 'NOT SIGNIFICANT'} at 95% confidence level")
```

```python
print("=" * 60)
print("STATISTICAL VALIDATION SUMMARY")
print("=" * 60)

print("\n1. Chi-Square: Loan Grade vs Default")
print(f"   Chi2 = 91,498.08 | df = 6 | p < 0.0001")
print(f"   Result: SIGNIFICANT — grade is a strong predictor of default")

print("\n2. Chi-Square: Loan Purpose vs Default")
print(f"   Chi2 = 4,093.66 | df = 13 | p < 0.0001")
print(f"   Result: SIGNIFICANT — purpose significantly affects default rate")

print("\n3. T-Test: Annual Income — Defaulted vs Paid")
print(f"   Defaulted mean: $70,363.87 | Paid mean: $77,706.18")
print(f"   T = -50.9510 | p < 0.0001")
print(f"   Result: SIGNIFICANT — defaulted borrowers earn $7,342 less on average")

print("\n4. T-Test: DTI — Defaulted vs Paid")
print(f"   Defaulted mean DTI: 20.20 | Paid mean DTI: 17.81")
print(f"   T = 94.6517 | p < 0.0001")
print(f"   Result: SIGNIFICANT — defaulted borrowers carry 2.39 higher DTI on average")

print("\n" + "=" * 60)
print("All 4 tests significant at p < 0.0001")
print("=" * 60)
```
